# idd-figures painters

A **painter** draws on a given `ax` and returns it — it never makes a Figure, never saves, and never loads data. A **layout** owns the Figure and calls painters (see `02_layouts`).

This vignette is *import-only*: it imports painters and reference exemplars from `idd_figures.lib` and calls them on **made-up, seeded data**. Domain differences (which columns, labels, colours) are just parameters.

```python
import matplotlib.pyplot as plt
from idd_figures.lib import example_data as ed, examples as ex
from idd_figures.lib.painters.lines import lines_panel
```

## lines
The canonical atom: one or more series over an ordered x, with an optional `[lo, hi]` band and anchor vline.

```python
ex.exemplar_lines()
```

Call the painter directly with a small config dict — the recommended call style. `value_scale='auto'` picks a readable multiplier and appends the unit to the y-label.

```python
df = ed.make_timeseries_df(n_series=3)
cfg = {'x': 'year_id', 'value': 'value', 'lo': 'lo', 'hi': 'hi', 'hue': 'series'}
fig, ax = plt.subplots(figsize=(8, 5))
lines_panel(ax, df, **cfg, value_scale='auto', ylabel='value')
fig
```

## Altering the defaults (these are *painter* features)

Per-mark appearance — **line colour, shaded-region colour, thickness (`lw`), transparency (`alpha` / `band_alpha`), line shape (`linestyle`), and markers** — are **painter** features. Each knob takes either one value (applied to every series) or a **dict keyed by series**.

(Fonts, tick placement, rounding, titles, and where panels sit are **layout / caller** features — set on the axes/figure or via `rcParams`. See `02_layouts`.)

```python
df = ed.make_timeseries_df(n_series=3)
cfg = {'x': 'year_id', 'value': 'value', 'lo': 'lo', 'hi': 'hi', 'hue': 'series'}
fig, ax = plt.subplots(figsize=(8, 5))
lines_panel(
    ax, df, **cfg,
    colors={'series_0': 'crimson', 'series_1': 'darkgreen', 'series_2': 'navy'},      # line colour
    band_color={'series_0': 'lightpink', 'series_1': 'lightgreen', 'series_2': 'lightblue'},  # band colour
    band_alpha=0.35,                                                                  # band transparency
    lw=2.5,                                                                           # thickness
    linestyle={'series_0': '-', 'series_1': '--', 'series_2': ':'},                    # line shape
    alpha=0.9,                                                                        # line transparency
    marker={'series_0': 'o'}, markersize=4,                                           # markers on one series
    xlabel='year', ylabel='value',
)
fig
```

## scatter
Size encoding + reference lines. The national unit is overlaid by calling the painter again on the same `ax`.

```python
ex.exemplar_scatter()
```

Same painter, different flags — and the **shaded-region colour/transparency** and marker colour are painter knobs too:

```python
from idd_figures.lib.painters.scatter import plot_scatter

lb = ed.make_left_behind_df()
ax = plot_scatter(
    lb, x='x', y='y', color='purple', alpha=0.6,
    shade={'x': (lb['x'].min(), 0), 'y': (lb['y'].min(), 0),
           'color': 'gold', 'alpha': 0.25, 'text': 'region'},
    ref_lines={'h': 0, 'v': 0}, xlabel='Δ x', ylabel='Δ y',
)
ax.figure
```
