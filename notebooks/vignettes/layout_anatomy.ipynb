# Layout anatomy — which number controls which space, and how to CHANGE it

Executable guides (curriculum item 2, `notebooks/vignettes/PLAN.md`): every band,
gap, and margin is annotated with the **kwarg name and its live value**, measured
from the realised figure — the guides cannot drift from the code. Every guide takes
its knobs as parameters, so each section shows the **default** call and then a
**changed** call — the diagram re-annotates itself with the new numbers. All figures
come from `idd_figures.lib.guides` (vignettes are import-only; no `def` here).

```python
%matplotlib inline
from idd_figures.lib import guides
```

## 1. The grid: margins, gaps, ratios

`margins` bound the panel area (figure fractions); `height_ratios`/`width_ratios`
split what's inside; `wspace`/`hspace` are charged against the **mean** panel size
(`geometry.gap_factor` inverts that charge). Nested grids have no margins of their
own — the parent's cell IS the margin control.

```python
guides.guide_grid_anatomy();
```

**Changed:** wide gutters, tight margins, equal rows, 3:1 columns — every label updates and the panels re-shape.

```python
guides.guide_grid_anatomy(
    wspace=0.8, hspace=0.15, height_ratios=(1, 1), width_ratios=(3, 1),
    margins={"left": 0.05, "right": 0.98, "top": 0.95, "bottom": 0.06},
    notes=False,  # tight margins leave no caption band
);
```

## 2. One map: stacked bands, derived height

Only the band heights are chosen; the map row is **derived** from `fig_width` and
the extent aspect. The aspect is explicit (`set_aspect(1.0, "box")`) and guarded —
geometry-breaking `margins`/`hspace` raise instead of stretching the map.

```python
guides.guide_map_panel_anatomy();
```

**Changed:** a SQUARE extent doubles the derived map row (8.0 × 4/4 = 8") — no height was touched; the extent did it. Title band doubled explicitly.

```python
guides.guide_map_panel_anatomy(extent=(0, 4, 0, 4), title_h=1.0);
```

## 3. map_facet: title-allowance gaps, bar cells, pads

Every inter-row gap IS `panel_title_h` — titles never survive on accidental slack.
Colorbars are grid cells spanning the panels they serve; row heights come from
`geometry.map_row_height`. (Rendered in preview mode on the synthetic fixture.)

```python
guides.guide_map_facet_anatomy();
```

**Changed:** `panel_title_h` more than doubled — every gap widens by exactly that amount (the figure grows to pay for it; nothing is stolen from the maps). Wider `wspace` narrows each panel, which shrinks the DERIVED row height.

```python
guides.guide_map_facet_anatomy(panel_title_h=0.9, wspace=0.15);
```

## 4. Inside a legend/colorbar cell

The cell's grid span is the *footprint*; these knobs set the drawn fill within it
(fractions OF THE CELL). A bar's ink — ticks, `cbar_label` — must stay inside its
own cell. Annotations are measured from the drawn swatches.

```python
guides.guide_bar_cell_anatomy();
```

**Changed:** fat side margins, wide swatch spacing, a low thin band — same cell, same painter, four numbers.

```python
guides.guide_bar_cell_anatomy(margin=0.20, spacing=0.08, bin_bottom=0.30, bin_top=0.60);
```

## 5. Text hangs outside its box

The solver places BOXES; titles, tick labels and axis labels render outside them,
so each needs an explicit band (`show_anatomy` overlay labels the regions). Text
extents exist only after a draw — hence the post-draw clipping guard on the backlog.

```python
guides.guide_text_placement();
```

**Changed (a failure on purpose):** tighten `left` and `bottom` and the ylabel/tick regions crowd the dotted figure boundary — the band was the only thing keeping that ink inside the canvas. This is the clipping bug from the mixed gallery figure, reproduced deliberately.

```python
guides.guide_text_placement(
    margins={"left": 0.07, "right": 0.95, "top": 0.86, "bottom": 0.14}, notes=False
);
```

## 6. One point, many addresses

The same letter placed via `transAxes` vs figure coordinates: when the layout
changes (bottom row), the axes-frame letter follows its panel; the figure-frame
letter stays glued to the canvas. All frames interconvert exactly, so anything can
be forced anywhere — explicitly.

```python
guides.guide_coordinate_frames();
```

**Changed:** the address moves to the panel centre — BOTH placements (and their printed coordinates) follow, because the letter's position was never a magic number, just an address in a declared frame.

```python
guides.guide_coordinate_frames(letter_xy=(0.5, 0.55));
```

## 7. Fonts: every text thing has an owner

There is no single per-region "font" knob. Each text object answers to its own setter — `set_title`, `set_xlabel`, `tick_params`, `suptitle` — shown here with deliberately mismatched sizes and the exact call labeled beside the plot. Wholesale defaults (family, base size) go through `mpl.rc_context({...})` around figure creation. Note what `margins` does NOT do: it moves the box, never the fonts.

```python
guides.guide_text_styles();
```

**Changed:** the title shrinks to normal 11pt, the xlabel balloons to 24, y ticks drop to 7 — every label in the call column updates, because each text object answered to its own setter.

```python
guides.guide_text_styles(title_size=11, xlabel_size=24, ytick_size=7, title_weight="normal");
```
